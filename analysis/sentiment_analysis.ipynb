```python
import pandas as pd
import pprint
import matplotlib.pyplot as plt
from HanziNLP import sentiment
```

```python
speeches = pd.read_csv(
    "/home/lburton12/side_projects/xijinpin_talks/corpus/speeches.csv"
)
```

```python
def filtering():
    speeches["publication"] = pd.to_datetime(speeches["publication"])

    speeches_2022 = speeches[speeches["publication"].dt.year == 2022]
    speeches_2023 = speeches[speeches["publication"].dt.year == 2023]
    speeches_2024 = speeches[speeches["publication"].dt.year == 2024]
    speeches_2025 = speeches[speeches["publication"].dt.year == 2025]

    # Looking for Speeches mentioning Taiwan

    return speeches_2022, speeches_2023, speeches_2024, speeches_2025

speeches_2022, speeches_2023, speeches_2024, speeches_2025 = filtering()
```

```python
#Sentiment Analysis
limit = 512

def sentiment_analysis(speeches,year):

    all_speeches_scores = []
    sentiment_averages = {}

    #Gathering all sentiment scores
    for index, text in speeches['text'].items():
        result = sentiment(text[:limit], model='touch20032003/xuyuan-trial-sentiment-bert-chinese',show = False)
        all_speeches_scores.append(result)

    for results in all_speeches_scores:

        for label,scores in results.items():

            if label not in sentiment_averages:
                sentiment_averages[label] = 0
            sentiment_averages[label] += scores
    
    for label in sentiment_averages:
        sentiment_averages[label] = sentiment_averages[label] / len(all_speeches_scores)

    #Plotting results
    fig, ax = plt.subplots()
    ax.bar(sentiment_averages.keys(),sentiment_averages.values())
    plt.title(f"Xi Jinping speeches in {year}: sentiment score averages")

    return plt.show()
                
```

```python
sentiment_analysis(speeches_2022,2022)
```

```python
sentiment_analysis(speeches_2023,2023)
```

```python
sentiment_analysis(speeches_2024,2024)
```

```python
sentiment_analysis(speeches_2025,2025)
```
